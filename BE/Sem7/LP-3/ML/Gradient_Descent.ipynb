Machine Learning Gradient Descent Algorithm

Implement Gradient Descent Algorithm to find the local minima 
of a function. 
For example, find the local minima of the function 
y=(x+3)² starting from the point 
x=2

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Objective function: y = (x + 3)^2
def objective(x):
    return (x + 3)**2
```

```python
# Derivative: dy/dx = 2(x + 3)
def derivative(x):
    return 2 * (x + 3)
```

```python
# Gradient Descent function
def gradient_descent(alpha, start, max_iter):
    """
    Performs gradient descent to find the local minimum.
    alpha      : learning rate
    start      : initial x value
    max_iter   : number of iterations
    """
    x = start
    history = []

    for i in range(max_iter):
        grad = derivative(x)
        x_new = x - alpha * grad
        y_new = objective(x_new)
        history.append([i+1, x, grad, x_new, y_new])
        x = x_new

    # Convert to DataFrame for table display
    df = pd.DataFrame(history, columns=["Iteration", "x_old", "Gradient", "x_new", "f(x_new)"])
    return df, x_new
```

```python
# --- Parameters ---
alpha = 0.1        # learning rate
start = 2          # starting point
max_iter = 50      # number of iterations for display
```

```python
# --- Run Gradient Descent ---
df, final_x = gradient_descent(alpha, start, max_iter)
```

```python
print(df)
print(f"\nLocal minimum approximately at x = {final_x:.6f}, f(x) = {objective(final_x):.6f}")
```

```python
# --- Plot the function and the GD path ---
x_vals = np.linspace(-6, 4, 100)
y_vals = objective(x_vals)

plt.plot(x_vals, y_vals, label="y = (x + 3)^2")
plt.title("Gradient Descent Path")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid(True)
plt.show()
```
